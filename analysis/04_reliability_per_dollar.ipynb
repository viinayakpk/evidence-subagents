Draft reliability-per-dollar notebook. Mock costs are zero when traces exist.

```python
from pathlib import Path
import json
folder = Path("runs/mock_e2e")
cost = 0.0
passed = 0
graded = 0
n = 0
if folder.is_dir():
    for path in sorted(folder.glob("*.jsonl")):
        for line in path.read_text().splitlines():
            event = json.loads(line)
            if event.get("type") == "run_end":
                n += 1
                # Agent calls and L3 judge calls are logged separately; both are spend.
                cost += float(event.get("cost_usd") or 0) + float(event.get("judge_cost_usd") or 0)
                # passed is null for runs whose environment did not start; they are not failures.
                if event.get("passed") is not None:
                    graded += 1
                    passed += int(bool(event["passed"]))
print({"runs": n, "graded": graded, "passed": passed, "cost_usd": cost})
```

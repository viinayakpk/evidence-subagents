Draft headline notebook. Compares arms once traces exist.

```python
from pathlib import Path
import json
from evcarry.eval.metrics import pass_rate
rows = []
for folder in (Path("runs/mock_e2e"), Path("runs/headline_deepseek_v1")):
    if not folder.is_dir():
        continue
    for path in folder.glob("*.jsonl"):
        for line in path.read_text().splitlines():
            event = json.loads(line)
            if event.get("type") == "run_end":
                rows.append(event)
if not rows:
    print("no traces yet")
else:
    for arm in sorted({row["arm"] for row in rows}):
        print(arm, pass_rate(rows, arm))
```

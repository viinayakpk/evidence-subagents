Draft headline notebook. Compares arms once traces exist.

```python
from pathlib import Path
import json
from evcarry.eval.metrics import pass_rate
# One experiment per table: pooling runs/mock_e2e (scripted, always passes) with the
# headline traces would inflate every arm's pass rate.
folder = Path("runs/headline_deepseek_v1")
rows = []
if folder.is_dir():
    for path in folder.glob("*.jsonl"):
        for line in path.read_text().splitlines():
            event = json.loads(line)
            if event.get("type") == "run_end":
                rows.append(event)
if not rows:
    print("no traces yet")
else:
    for arm in sorted({row["arm"] for row in rows}):
        graded = [row for row in rows if row["arm"] == arm and row.get("passed") is not None]
        print(arm, pass_rate(rows, arm), f"graded={len(graded)}")
```

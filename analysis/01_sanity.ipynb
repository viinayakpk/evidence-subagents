Draft sanity notebook. Reads JSONL traces when a mock run directory exists.

```python
from pathlib import Path
import json
folder = Path("runs/mock_e2e")
rows = []
if folder.is_dir():
    for path in sorted(folder.glob("*.jsonl")):
        for line in path.read_text().splitlines():
            event = json.loads(line)
            if event.get("type") == "run_end":
                rows.append(event)
print(len(rows), "run_end rows")
```
